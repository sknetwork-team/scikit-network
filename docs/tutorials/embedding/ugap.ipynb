# UGAP

This notebook illustrates the embedding of a graph with UGAP, based on Personalized PageRank similarity.



```python
from IPython.display import SVG
```

```python
from sknetwork.data import karate_club
from sknetwork.embedding import UGAP
from sknetwork.visualization import visualize_graph
```

## Example

```python
graph = karate_club(metadata=True)
adjacency = graph.adjacency
labels = graph.labels
```

```python
ugap = UGAP(n_components=2, random_state=42)
embedding = ugap.fit_transform(adjacency)
image = visualize_graph(adjacency, embedding, labels=labels)
SVG(image)
```
